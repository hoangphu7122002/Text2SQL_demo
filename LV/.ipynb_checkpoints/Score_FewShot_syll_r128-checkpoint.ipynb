```python
import os
os.environ["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
os.environ["CUDA_VISIBLE_DEVICES"] = "4"
```

```python
filenames = os.listdir("benchmark__FewShot_codeLlama_text2sql_syll_r128_2")
filenames[:5]
```

```python
import json

dataset = []
for filename in filenames:
    if not filename.endswith(".json"):
        continue
    with open("benchmark__FewShot_codeLlama_text2sql_syll_r128_2/" + filename) as _f:
        data = json.load(_f)
    dataset.append(data)
```

```python
from datasets import Dataset

dataset = Dataset.from_list(dataset)
```

```python
dataset.push_to_hub("TeeA/VinAIResearch-ViText2SQL",config_name="benchmark__FewShot_codeLlama_text2sql_syll_r128",data_dir="benchmark/benchmark__FewShot_codeLlama_text2sql_syll_r128")
```
